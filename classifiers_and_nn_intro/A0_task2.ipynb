# Task 2: Implement a multi-class perceptron algorithm

```python
import numpy as np
```

```python
train_in = np.loadtxt('train_in.csv', delimiter=',')
train_out = np.loadtxt('train_out.csv', delimiter=',')
test_in = np.loadtxt('test_in.csv', delimiter=',')
test_out = np.loadtxt('test_out.csv', delimiter=',')
```

```python
num_classes = 10 
learning_rate = 0.01
num_epochs = 250
```

```python
input_dim = train_in.shape[1]
X = np.append(train_in, np.ones((train_in.shape[0], 1)), axis=1)
print(X.shape)
W = np.random.randn(input_dim + 1, num_classes) 
W.shape
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))
```

```python
# Training the perceptron
delta = np.zeros(num_classes)
delta.shape
for _ in range(num_epochs):
    
    for j in range(train_in.shape[0]):
        x = X[j]
        y_true = train_out[j]

        y_pred = np.dot(x, W)
        output_layer = sigmoid(y_pred)

        for i in range(num_classes):
            if i == y_true:
                delta[i] = sigmoid_derivative(output_layer[i]) * (1 - output_layer[i])
            else:
                delta[i] = sigmoid_derivative(output_layer[i]) * (0 - output_layer[i])
    
        W += learning_rate * np.dot(x.reshape(input_dim + 1, 1), delta.reshape(num_classes, 1).T)
   
    correct = 0
    total = train_in.shape[0]
    X_train = np.append(train_in, np.ones((train_in.shape[0], 1)), axis=1)

    for i in range(total):
        x = X_train[i]
        y_true = int(train_out[i])
        y_pred = np.dot(x, W)
        predicted_class = sigmoid(y_pred)
        predicted_class = np.argmax(predicted_class)
        if int(predicted_class) == int(y_true):
            correct += 1

    print("Epoch {}: {}".format(_, correct ))
    
```

```python
# Testing the perceptron on the train set
correct = 0
total = train_in.shape[0]
X_train = np.append(train_in, np.ones((train_in.shape[0], 1)), axis=1)

for i in range(total):
    x = X_train[i]
    y_true = int(train_out[i])
    y_pred = np.dot(x, W)
    predicted_class = sigmoid(y_pred)
    predicted_class = np.argmax(predicted_class)
    if int(predicted_class) == int(y_true):
        correct += 1

accuracy = correct / total
print("Train Accuracy:", accuracy)
```

```python
# Testing the perceptron on the test set
correct = 0
total = test_in.shape[0]
X_test = np.append(test_in, np.ones((test_in.shape[0], 1)), axis=1)

for i in range(total):
    x = X_test[i]
    y_true = int(test_out[i])
    y_pred = np.dot(x, W)
    predicted_class = sigmoid(y_pred)
    predicted_class = np.argmax(predicted_class)

    if predicted_class == y_true:
        correct += 1

accuracy = correct / total
print("Test Accuracy:", accuracy)
```
